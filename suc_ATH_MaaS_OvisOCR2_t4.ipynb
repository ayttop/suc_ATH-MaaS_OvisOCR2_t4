```python
import os
import io
import re
import time
import base64
import threading
from pathlib import Path
from typing import Iterator, Any

import torch
import fitz
import gradio as gr

from PIL import Image, ImageOps
from transformers import (
    AutoProcessor,
    Qwen3_5ForConditionalGeneration,
    TextIteratorStreamer,
)

# ============================================================
# CONFIG
# ============================================================

MODEL_ID = "ATH-MaaS/OvisOCR2"

MAX_NEW_TOKENS = 16384

# نفس الفكرة الموجودة في Space الرسمي
MIN_PIXELS = 448 * 448
MAX_PIXELS = 2880 * 2880

PDF_RENDER_SCALE = 2.0
MAX_PDF_PAGES = 50

# ============================================================
# CUDA / MEMORY
# ============================================================

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
os.environ["TOKENIZERS_PARALLELISM"] = "false"

if not torch.cuda.is_available():
    raise RuntimeError(
        "CUDA غير متاحة. فعّل T4 من Runtime > Change runtime type."
    )

print("=" * 60)
print("OvisOCR2 - Google Colab T4")
print("=" * 60)
print("PyTorch :", torch.__version__)
print("CUDA    :", torch.version.cuda)
print("GPU     :", torch.cuda.get_device_name(0))
print(
    "VRAM    :",
    round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2),
    "GB",
)
print("=" * 60)


# ============================================================
# LOAD MODEL
# ============================================================

print("\nLoading OvisOCR2...")

processor = AutoProcessor.from_pretrained(
    MODEL_ID,
    min_pixels=MIN_PIXELS,
    max_pixels=MAX_PIXELS,
)

model = Qwen3_5ForConditionalGeneration.from_pretrained(
    MODEL_ID,
    dtype=torch.bfloat16,
    attn_implementation="sdpa",
)

model = model.to("cuda")
model.eval()

print("\nOvisOCR2 loaded successfully!")
print("Device:", model.device)
print("=" * 60)


# ============================================================
# OCR PROMPT
# ============================================================

OCR_PROMPT = """
Extract all readable content from the image in natural human reading order
and output the result as a single Markdown document.

For charts or images, represent them using an HTML image tag:
<img src="images/bbox_{left}_{top}_{right}_{bottom}.jpg" />

where left, top, right, bottom are bounding box coordinates scaled to [0, 1000).

Format formulas as LaTeX.

Format tables as HTML:
<table>...</table>

Transcribe all other text as standard Markdown.

Preserve the original text without translation or paraphrasing.
""".strip()


# ============================================================
# HELPERS
# ============================================================

def get_file_path(file_obj):
    """
    Get actual uploaded file path from Gradio.
    Supports both string paths and Gradio file objects.
    """

    if file_obj is None:
        raise ValueError("لم يتم رفع أي ملف.")

    if isinstance(file_obj, str):
        return file_obj

    if isinstance(file_obj, dict):
        path = file_obj.get("path")
        if path:
            return str(path)

    path = getattr(file_obj, "path", None)

    if path:
        return str(path)

    raise ValueError("تعذر الحصول على مسار الملف المرفوع.")


def document_info(path):
    """
    Detect whether uploaded file is an image or PDF.
    """

    path = str(path)
    suffix = Path(path).suffix.lower()

    # Check PDF header
    try:
        with open(path, "rb") as f:
            header = f.read(5)
    except Exception as e:
        raise ValueError("تعذر قراءة الملف.") from e

    if suffix == ".pdf" or header == b"%PDF-":

        with fitz.open(path) as doc:
            total_pages = doc.page_count

        if total_pages < 1:
            raise ValueError("ملف PDF لا يحتوي على صفحات.")

        if total_pages > MAX_PDF_PAGES:
            raise ValueError(
                f"الحد الأقصى هو {MAX_PDF_PAGES} صفحة. "
                f"الملف يحتوي على {total_pages} صفحة."
            )

        return "pdf", total_pages

    # Validate image
    try:
        with Image.open(path) as img:
            img.verify()
    except Exception as e:
        raise ValueError(
            "الملف يجب أن يكون PNG أو JPEG أو WebP أو PDF."
        ) from e

    return "image", 1


def load_page(path, document_type, page_index):
    """
    Load one page as PIL RGB image.
    """

    if document_type == "pdf":

        with fitz.open(path) as doc:

            page = doc.load_page(page_index)

            pix = page.get_pixmap(
                matrix=fitz.Matrix(
                    PDF_RENDER_SCALE,
                    PDF_RENDER_SCALE,
                ),
                colorspace=fitz.csRGB,
                alpha=False,
            )

            return Image.frombytes(
                "RGB",
                (pix.width, pix.height),
                pix.samples,
            )

    with Image.open(path) as img:
        return ImageOps.exif_transpose(img).convert("RGB")


def generation_token_ids():
    tokenizer = processor.tokenizer

    return {
        "eos_token_id": int(tokenizer.eos_token_id),
        "pad_token_id": int(tokenizer.pad_token_id),
    }


# ============================================================
# MODEL INPUT
# ============================================================

def create_inputs(image):
    """
    Create OvisOCR2 multimodal inputs.
    """

    messages = [
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "image": image,
                },
                {
                    "type": "text",
                    "text": OCR_PROMPT,
                },
            ],
        }
    ]

    inputs = processor.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_dict=True,
        return_tensors="pt",
        enable_thinking=False,
    )

    return inputs.to(model.device)


# ============================================================
# OCR STREAM
# ============================================================

def run_ocr_stream(image):
    """
    Run OCR and stream text progressively.
    """

    inputs = create_inputs(image)

    streamer = TextIteratorStreamer(
        processor.tokenizer,
        skip_prompt=True,
        skip_special_tokens=True,
        clean_up_tokenization_spaces=False,
    )

    errors = []

    def generate():

        try:

            with torch.inference_mode():

                model.generate(
                    **inputs,
                    streamer=streamer,
                    max_new_tokens=MAX_NEW_TOKENS,
                    do_sample=False,
                    temperature=None,
                    top_p=None,
                    top_k=None,
                    **generation_token_ids(),
                )

        except Exception as e:

            errors.append(e)

            try:
                streamer.on_finalized_text(
                    "",
                    stream_end=True,
                )
            except Exception:
                pass

    worker = threading.Thread(
        target=generate,
        daemon=True,
    )

    worker.start()

    text = ""

    for fragment in streamer:

        text += fragment

        yield text

    worker.join()

    if errors:
        raise RuntimeError(
            f"فشل تشغيل OvisOCR2: {errors[0]}"
        )

    return text


# ============================================================
# BBOX MATERIALIZATION
# ============================================================

BBOX_PATTERN = re.compile(
    r'<img\s+src=["\']images/bbox_'
    r'(\d+)_(\d+)_(\d+)_(\d+)\.jpg["\']\s*/?>',
    flags=re.IGNORECASE,
)


def materialize_bbox_images(markdown, page_image):
    """
    Replace bbox references produced by OvisOCR2
    with embedded base64 JPEG crops.
    """

    if not markdown:
        return markdown

    width, height = page_image.size

    def replace(match):

        left, top, right, bottom = (
            int(x) for x in match.groups()
        )

        x1 = max(
            0,
            min(
                width,
                round(left * width / 1000),
            ),
        )

        y1 = max(
            0,
            min(
                height,
                round(top * height / 1000),
            ),
        )

        x2 = max(
            0,
            min(
                width,
                round(right * width / 1000),
            ),
        )

        y2 = max(
            0,
            min(
                height,
                round(bottom * height / 1000),
            ),
        )

        if x2 <= x1 or y2 <= y1:
            return match.group(0)

        crop = page_image.crop(
            (x1, y1, x2, y2)
        ).convert("RGB")

        crop.thumbnail(
            (1200, 1200),
            Image.Resampling.BILINEAR,
        )

        buffer = io.BytesIO()

        crop.save(
            buffer,
            format="JPEG",
            quality=85,
        )

        encoded = base64.b64encode(
            buffer.getvalue()
        ).decode("ascii")

        return (
            '<img '
            f'src="data:image/jpeg;base64,{encoded}" '
            'alt="Visual region" '
            'style="max-width:100%;height:auto;" />'
        )

    return BBOX_PATTERN.sub(
        replace,
        markdown,
    )


# ============================================================
# SINGLE IMAGE OCR
# ============================================================

def ocr_image(file_obj):

    if file_obj is None:
        return "", "ارفع صورة أولًا."

    start = time.time()

    try:

        path = get_file_path(file_obj)

        document_type, total_pages = document_info(path)

        if document_type != "image":
            return "", "هذا زر الصور فقط. استخدم زر PDF."

        image = load_page(
            path,
            "image",
            0,
        )

        print(
            f"[OCR] Image size: {image.size}",
            flush=True,
        )

        final_text = ""

        for partial in run_ocr_stream(image):

            final_text = partial

            # Gradio يحتاج return في نهاية الوظيفة،
            # لذلك لا نستخدم yield هنا.
            # يتم تحديث النتيجة بعد انتهاء التوليد.

        rendered = materialize_bbox_images(
            final_text,
            image,
        )

        elapsed = time.time() - start

        status = (
            f"تم الانتهاء.\n"
            f"الوقت: {elapsed:.1f} ثانية\n"
            f"عدد الأحرف: {len(final_text):,}"
        )

        return rendered, status

    except Exception as e:

        return (
            "",
            f"ERROR: {type(e).__name__}: {e}",
        )


# ============================================================
# PDF OCR
# ============================================================

def ocr_pdf(file_obj, start_page, page_count):

    if file_obj is None:
        return "", "ارفع PDF أولًا."

    start_time = time.time()

    try:

        path = get_file_path(file_obj)

        document_type, total_pages = document_info(path)

        if document_type != "pdf":
            return "", "الملف ليس PDF."

        start_page = int(start_page) - 1
        page_count = int(page_count)

        if start_page < 0:
            start_page = 0

        if start_page >= total_pages:
            return (
                "",
                f"الصفحة المطلوبة غير موجودة. "
                f"الملف يحتوي على {total_pages} صفحات.",
            )

        end_page = min(
            total_pages,
            start_page + page_count,
        )

        all_pages = []

        print(
            f"[PDF] Processing pages "
            f"{start_page + 1}-{end_page}/{total_pages}",
            flush=True,
        )

        for page_index in range(
            start_page,
            end_page,
        ):

            page_number = page_index + 1

            print(
                f"[OCR] Page {page_number}/{end_page}",
                flush=True,
            )

            image = load_page(
                path,
                "pdf",
                page_index,
            )

            final_text = ""

            for partial in run_ocr_stream(image):
                final_text = partial

            rendered = materialize_bbox_images(
                final_text,
                image,
            )

            all_pages.append(
                f"<!-- Page {page_number} -->\n\n"
                f"{rendered}"
            )

            # تنظيف الذاكرة بعد كل صفحة
            del image

            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        result = "\n\n---\n\n".join(
            all_pages
        )

        elapsed = time.time() - start_time

        status = (
            f"تم الانتهاء.\n"
            f"الصفحات: {start_page + 1}-{end_page}\n"
            f"إجمالي الصفحات في PDF: {total_pages}\n"
            f"الوقت: {elapsed:.1f} ثانية\n"
            f"عدد الأحرف: {len(result):,}"
        )

        return result, status

    except Exception as e:

        return (
            "",
            f"ERROR: {type(e).__name__}: {e}",
        )


# ============================================================
# PDF INFO
# ============================================================

def pdf_info(file_obj):

    if file_obj is None:
        return "لم يتم اختيار PDF."

    try:

        path = get_file_path(file_obj)

        document_type, total_pages = document_info(path)

        if document_type != "pdf":
            return "الملف ليس PDF."

        return (
            f"PDF صحيح.\n"
            f"عدد الصفحات: {total_pages}"
        )

    except Exception as e:

        return f"ERROR: {e}"


# ============================================================
# GRADIO UI
# ============================================================

CUSTOM_CSS = """
#title {
    text-align: center;
    margin-bottom: 10px;
}

.status {
    white-space: pre-wrap;
}

textarea {
    font-family: monospace !important;
}

.output-markdown {
    min-height: 500px;
}
"""


with gr.Blocks(
    title="OvisOCR2 - Colab T4",
    theme=gr.themes.Soft(),
    css=CUSTOM_CSS,
) as demo:

    gr.Markdown(
        """
        # 🔎 OvisOCR2 OCR

        ### تشغيل OvisOCR2 محليًا على Google Colab T4

        النموذج يعمل بالكامل داخل جلسة Colab.
        """
        ,
        elem_id="title",
    )

    with gr.Tabs():

        # ====================================================
        # IMAGE TAB
        # ====================================================

        with gr.Tab("📷 صورة"):

            with gr.Row():

                with gr.Column():

                    image_input = gr.File(
                        label="ارفع صورة",
                        file_types=[
                            ".png",
                            ".jpg",
                            ".jpeg",
                            ".webp",
                        ],
                        type="filepath",
                    )

                    image_button = gr.Button(
                        "🚀 تشغيل OCR",
                        variant="primary",
                    )

                    image_status = gr.Textbox(
                        label="الحالة",
                        interactive=False,
                        elem_classes="status",
                    )

                with gr.Column():

                    image_output = gr.Markdown(
                        label="OCR Result",
                        elem_classes="output-markdown",
                    )

            image_button.click(
                fn=ocr_image,
                inputs=image_input,
                outputs=[
                    image_output,
                    image_status,
                ],
            )

        # ====================================================
        # PDF TAB
        # ====================================================

        with gr.Tab("📄 PDF"):

            with gr.Row():

                with gr.Column():

                    pdf_input = gr.File(
                        label="ارفع PDF",
                        file_types=[".pdf"],
                        type="filepath",
                    )

                    pdf_info_button = gr.Button(
                        "🔍 معلومات PDF",
                    )

                    pdf_info_output = gr.Textbox(
                        label="معلومات الملف",
                        interactive=False,
                    )

                    with gr.Row():

                        pdf_start = gr.Number(
                            label="صفحة البداية",
                            value=1,
                            minimum=1,
                            precision=0,
                        )

                        pdf_count = gr.Number(
                            label="عدد الصفحات",
                            value=1,
                            minimum=1,
                            maximum=50,
                            precision=0,
                        )

                    pdf_button = gr.Button(
                        "🚀 تشغيل OCR على PDF",
                        variant="primary",
                    )

                    pdf_status = gr.Textbox(
                        label="الحالة",
                        interactive=False,
                        elem_classes="status",
                    )

                with gr.Column():

                    pdf_output = gr.Markdown(
                        label="OCR Result",
                        elem_classes="output-markdown",
                    )

            pdf_info_button.click(
                fn=pdf_info,
                inputs=pdf_input,
                outputs=pdf_info_output,
            )

            pdf_button.click(
                fn=ocr_pdf,
                inputs=[
                    pdf_input,
                    pdf_start,
                    pdf_count,
                ],
                outputs=[
                    pdf_output,
                    pdf_status,
                ],
            )

    gr.Markdown(
        """
        ---
        **Model:** `ATH-MaaS/OvisOCR2`

        **Backend:** Transformers

        **GPU:** NVIDIA Tesla T4
        """
    )


# ============================================================
# LAUNCH
# ============================================================

print("\nStarting Gradio...")

demo.launch(
    share=True,
    debug=True,
    show_error=True,
)
```
